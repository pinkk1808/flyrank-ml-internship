## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Unit of analysis: One row represents one content page. Each row contains observable information about that page, such as its age, search impressions, CTR, average position, and trend direction. I will use these page-level features to rank which pages may deserve review first.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/pinkk1808/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

# Proxy target: 1 = declining page, 0 = not declining
df["needs_review_proxy"] = (df["trend_direction"] == "down").astype(int)

lane_df = df[
    [
        "content_age_days",
        "impressions_90d",
        "ctr",
        "avg_position",
        "trend_direction",
        "needs_review_proxy"
    ]
]

print("One row = one content page")
print("Shape:", lane_df.shape)

lane_df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML beats a fixed rule: A fixed rule might say that any old page with low CTR or a poor average position should be reviewed, but real pages can have many different combinations of signals. For example, an older page may still perform well, while a newer page may already be declining. ML can learn patterns across multiple features at the same time, such as content age, impressions, CTR, and average position, instead of relying on one hard threshold. This can make the ranking more flexible and better suited to prioritizing pages for human review.
