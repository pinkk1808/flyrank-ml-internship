## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Quick look at the data: The starter dataset contains 30,000 content pages. Among them, 16,262 pages (54.21%) are marked as declining. Also, 8,739 declining pages are at least 180 days old. These numbers suggest that there are many pages that may deserve review, so a ranking system could help a content or SEO team prioritize which pages to inspect first instead of treating all pages equally.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/pinkk1808/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

declining = df["trend_direction"].eq("down")

print("Total content pages:", len(df))
print("Declining pages:", declining.sum())
print("Declining percentage:", round(declining.mean() * 100, 2), "%")
print(
    "Declining pages aged 180+ days:",
    (declining & (df["content_age_days"] >= 180)).sum()
)
```
